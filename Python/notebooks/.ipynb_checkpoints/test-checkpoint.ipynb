```python
a = -2
b = 1
s = 0.1
def temp(t, y): return 2 * a * y  - b*b*s 
from scipy.integrate import solve_ivp
```

```python
q0 = -0.5
t0 = 0
tf = 10
sol = solve_ivp(temp, [t0, tf], [q0])
```

```python
t = sol.t
q = np.exp(2*a*t)*q0 - b*b*s*(t-t0)
```

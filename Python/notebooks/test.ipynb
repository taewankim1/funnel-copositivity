```python
num1 = 500
num2 = 2
a = np.random.randn(num1,num2,2*num2)
b = np.random.randn(num1,2*num2,num2)
begin = time.time()
c = a@b
print("parallel",time.time() - begin)
c_ = np.zeros_like(c)
begin = time.time()
for i in range(num1) :
    c_[i] = a[i]@b[i]
print("serial",time.time() - begin)
print(np.abs(np.sum(c-c_)))
```
